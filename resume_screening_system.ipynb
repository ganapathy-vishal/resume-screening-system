```python
from abc import ABC, abstractmethod

class Candidate(ABC):

    def __init__(self, name, skills, cgpa):
        self.__name = name
        self.__skills = skills
        self.__cgpa = cgpa

    def get_name(self):
        return self.__name

    def get_skills(self):
        return self.__skills

    def get_cgpa(self):
        return self.__cgpa

    @abstractmethod
    def check_experience(self):
        pass


class FresherCandidate(Candidate):

    def __init__(self, name, skills, cgpa):
        super().__init__(name, skills, cgpa)

    def check_experience(self):
        return 0

class ExperiencedCandidate(Candidate):

    def __init__(self, name, skills, cgpa, experience):
        super().__init__(name, skills, cgpa)
        self.experience = experience

    def check_experience(self):
        return self.experience


class Job:

    def __init__(self, job_title, required_skills,
                 min_cgpa, min_experience):

        self.job_title = job_title
        self.required_skills = required_skills
        self.min_cgpa = min_cgpa
        self.min_experience = min_experience


class ResumeScreening:

    def __init__(self, job):
        self.job = job

    def screen_candidate(self, candidate):

        print("\n------------------------------")
        print("CANDIDATE SCREENING")
        print("------------------------------")

        try:
            # Candidate details
            print("Name:", candidate.get_name())
            print("Skills:", candidate.get_skills())
            print("CGPA:", candidate.get_cgpa())
            print("Experience:",
                  candidate.check_experience(), "years")

            # Skill matching
            candidate_skills = {
                skill.strip().lower()
                for skill in candidate.get_skills()
            }

            required_skills = {
                skill.strip().lower()
                for skill in self.job.required_skills
            }

            matched_skills = (
                candidate_skills & required_skills
            )

            if required_skills:
                percentage = (
                    len(matched_skills)
                    / len(required_skills)
                ) * 100
            else:
                percentage = 0

            # Eligibility checking
            skill_eligible = percentage >= 60

            cgpa_eligible = (
                candidate.get_cgpa() >= self.job.min_cgpa
            )

            experience_eligible = (
                candidate.check_experience()
                >= self.job.min_experience
            )

            # Screening result
            print("\nMatched Skills:", matched_skills)
            print("Skill Match:", round(percentage, 2), "%")
            print("CGPA Eligible:", cgpa_eligible)
            print("Experience Eligible:", experience_eligible)

            if (skill_eligible and cgpa_eligible
                    and experience_eligible):

                print("\nStatus: SHORTLISTED")

            else:
                print("\nStatus: REJECTED")

        except (ValueError, TypeError) as error:
            print("Error:", error)

try:

    print("===== RESUME SCREENING SYSTEM =====")

    # Job details
    job = Job(
        "Python Developer",
        ["Python", "SQL", "HTML", "OOP"],
        7.0,
        1
    )

    print("\nJOB REQUIREMENTS")
    print("Job Title:", job.job_title)
    print("Required Skills:", job.required_skills)
    print("Minimum CGPA:", job.min_cgpa)
    print("Minimum Experience:", job.min_experience)

    # Candidate input
    candidates = []

    number = int(input("\nEnter number of candidates: "))

    for i in range(number):

        print(f"\nEnter Candidate {i+1} Details")

        name = input("Enter Name: ")

        skills = input(
            "Enter Skills (comma separated): "
        ).split(",")

        cgpa = float(input("Enter CGPA: "))

        candidate_type = input(
            "Fresher or Experienced (F/E): "
        ).strip().upper()

        if not 0 <= cgpa <= 10:
            raise ValueError("CGPA must be between 0 and 10")

        if candidate_type == "F":

            candidate = FresherCandidate(
                name, skills, cgpa
            )

        elif candidate_type == "E":

            experience = float(
                input("Enter Experience in years: ")
            )

            if experience < 0:
                raise ValueError("Experience cannot be negative")

            candidate = ExperiencedCandidate(
                name, skills, cgpa, experience
            )

        else:
            raise ValueError("Invalid candidate type")

        candidates.append(candidate)

    # Screening
    screening = ResumeScreening(job)

    shortlisted = []
    rejected = []

    for candidate in candidates:

        screening.screen_candidate(candidate)

        # Store screening status
        candidate_skills = {
            skill.strip().lower()
            for skill in candidate.get_skills()
        }

        required_skills = {
            skill.strip().lower()
            for skill in job.required_skills
        }

        match_percentage = (
            len(candidate_skills & required_skills)
            / len(required_skills)
        ) * 100 if required_skills else 0

        if (
            match_percentage >= 60
            and candidate.get_cgpa() >= job.min_cgpa
            and candidate.check_experience() >= job.min_experience
        ):
            shortlisted.append(candidate.get_name())
        else:
            rejected.append(candidate.get_name())

    # Final result
    print("\n========== FINAL RESULT ==========")

    print("\nSHORTLISTED CANDIDATES:")
    for name in shortlisted:
        print("-", name)

    print("\nREJECTED CANDIDATES:")
    for name in rejected:
        print("-", name)

except (ValueError, TypeError) as error:
    print("\nInput Error:", error)
```
